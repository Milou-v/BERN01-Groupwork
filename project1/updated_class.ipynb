```python
#@njit
def calc_U_1_particle(con, p_idx, p_new, LENGTH,
                      sigma_0, sigma_1, epsilon):

    U = 0.0

    for j in range(len(con)):
        if j == p_idx:
            continue

        dx = con[j, 0] - p_new[0]
        dy = con[j, 1] - p_new[1]

        dx -= np.rint(dx / LENGTH) * LENGTH
        dy -= np.rint(dy / LENGTH) * LENGTH

        r2 = dx*dx + dy*dy

        if r2 < sigma_0**2:
            return np.inf

        if r2 < sigma_1**2:
            U += epsilon

    return U
```
